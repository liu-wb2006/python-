## 欧几里得算法

用辗转相除法（`a % b`）求两个整数 `m, n` 的最大公约数。

```python
def gcd(m, n):
    a, b = m, n
    while b != 0:
        a, b = b, a % b
    return a
```

### 计算下列数字对的 GCD

1. (2018, 2019)
2. (1200, 300)
3. (5040, 60)


```python
print("gcd(2018, 2019) = ", gcd(2018, 2019))
print("gcd(1200, 300)  = ", gcd(1200, 300))
print("gcd(5040, 60)   = ", gcd(5040, 60))
```

### 如果一个或两个参数是负数，函数会做什么？



```python
# 负数输入的行为演示
print("gcd(-2018, 2019) = ", gcd(-2018, 2019))
print("gcd(-1200, -300) = ", gcd(-1200, -300))
print("gcd(1200, -300)  = ", gcd(1200, -300))
```



函数本身没有任何特殊处理，仍然会照常运行，但**结果不再符合数学上“最大公约数是正数”的约定**：

- 例如 `gcd(-2018, 2019) = 1`，因为 `-2018 % 2019 = 1`，碰巧仍等于两者绝对值的最大公约数；
- 但如果两个数都是负数，例如 `gcd(-1200, -300) = -300`，会得到一个**负数**结果；
- `gcd(1200, -300)` 会得到 `-300`，因为当余数恰好为 0 时，返回的是被更新的除数 `-300`，符号随 `%` 运算的除数变化。

结论：这种行为**没有数学意义**。最大公约数按定义应为非负整数，更稳妥的做法是先取绝对值再计算，即 `gcd(abs(m), abs(n))`。

## 近似欧拉数 e





```python
import math


def euler_limit(n):
    # 公式 (2)：取 x = n
    return (1 + 1 / n) ** n


def euler_infinite_sum(n):
    # 公式 (1) 的前 n 项：k = 0, 1, ..., n-1（函数按 1-索引计数）
    total = 0.0
    for k in range(n):
        total += 1.0 / math.factorial(k)
    return total


def euler_approx(epsilon):
    # 用公式 (1) 求和，直到与真实值 e 的差在 epsilon 之内
    true_e = math.exp(1)
    total = 0.0
    k = 0
    while abs(total - true_e) > epsilon:
        total += 1.0 / math.factorial(k)
        k += 1
    return total


def print_euler_sum_table(n):
    # 打印 euler_infinite_sum(k)，k = 1 .. n，每行一个
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))
```

```python
# 测试：euler_infinite_sum 的索引约定
print("euler_infinite_sum(0) = ", euler_infinite_sum(0))  # 期望 0
print("euler_infinite_sum(1) = ", euler_infinite_sum(1))  # 期望 1
print("euler_infinite_sum(4) = ", euler_infinite_sum(4))  # 期望 ≈ 2.667
```

```python
# 测试：euler_limit 与 euler_approx
print("euler_limit(1_000_000)  = ", euler_limit(1_000_000))
print("euler_approx(1e-10)     = ", euler_approx(1e-10))
print("math.exp(1)（真实值 e） = ", math.exp(1))
```

```python
# 测试：print_euler_sum_table(10)
print_euler_sum_table(10)
```

## 测试整数属性

一个正整数 $n$ 是 2 的幂，如果 $n = 2^{p}$ 对某个整数 $p \geq 0$ 成立。

```python
def is_power_of_2(n):
    # 只使用除法和模运算判断 n 是否为 2 的幂
    while n > 1 and n % 2 == 0:
        n //= 2
    return n == 1
```

```python
# 测试 is_power_of_2
for n in [1, 2, 3, 4, 8, 16, 24, 32, 64]:
    print(f"is_power_of_2({n}) = {is_power_of_2(n)}")
```

```python
def is_power(b, n):
    # 返回 n 是否为 b 的幂（b, n 均为正整数）
    if b == 1:
        return n == 1  # 1 的任何次幂都是 1
    while n > 1 and n % b == 0:
        n //= b
    return n == 1
```

```python
# 测试 is_power
for b, n in [(2, 8), (2, 10), (3, 27), (3, 28), (5, 125), (4, 2), (1, 1), (1, 5)]:
    print(f"is_power({b}, {n}) = {is_power(b, n)}")
```
