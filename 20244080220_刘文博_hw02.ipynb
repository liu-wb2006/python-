## 一、字符串的乐趣

```python
def is_palindrome(s):
    """忽略空格和大小写，判断是否为回文。"""
    chars = [c.lower() for c in s if c != " "]
    return chars == chars[::-1]


print(is_palindrome("level"))
print(is_palindrome("TACO cat"))
print(is_palindrome("tacocat"))
print(is_palindrome("abc"))
```

```python
def is_abecedarian(s):
    """忽略空格，判断字母是否按字母顺序排列（允许重复）。"""
    letters = [c.lower() for c in s if c != " "]
    return all(letters[i] <= letters[i + 1] for i in range(len(letters) - 1))


print(is_abecedarian("adder"))
print(is_abecedarian("beet"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("dog"))
print(is_abecedarian("cat"))
```

```python
def remove_vowels(s):
    """去掉所有元音字母 a e i o u（不区分大小写）。"""
    return "".join(c for c in s if c.lower() not in "aeiou")


print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
```

## 二、列表的乐趣

```python
def list_reverse(lst):
    """返回反转后的列表；输入不是列表时抛出 TypeError。"""
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]


print(list_reverse([1, 2, 3]))
print(list_reverse(["a", "b"]))
```

```python
def is_sorted(seq):
    """只遍历一次，判断序列是否按升序排列。"""
    return all(seq[i] <= seq[i + 1] for i in range(len(seq) - 1))


print(is_sorted([1, 2, 3, 4]))
print(is_sorted([1, 3, 2]))
print(is_sorted((2, 2, 5)))
```

## 三、作为向量的元组

```python
def vec_scalar_mult(t, s):
    """向量（元组）数乘；类型不合适时抛出 TypeError。"""
    if not isinstance(t, tuple) or not all(isinstance(x, (int, float)) for x in t):
        raise TypeError("向量必须是由整数或浮点数组成的元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量必须是整数或浮点数")
    return tuple(x * s for x in t)


print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((1.5, -2), 3))
```

```python
def vec_inner_product(x, y):
    """两个向量的内积，返回浮点数。"""
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("两个向量都必须是元组")
    if len(x) != len(y):
        raise ValueError("两个向量的长度必须一致")
    return float(sum(a * b for a, b in zip(x, y)))


print(vec_inner_product((1, 2, 3), (4, 5, 6)))
```

```python
def check_valid_mx(mx):
    """判断是否为有效矩阵：外层元组、内层等长元组、元素均为整数或浮点数。"""
    if not isinstance(mx, tuple) or len(mx) == 0:
        return False
    if not all(isinstance(row, tuple) for row in mx):
        return False
    n = len(mx[0])
    if not all(len(row) == n for row in mx):
        return False
    return all(isinstance(x, (int, float)) for row in mx for x in row)


print(check_valid_mx(((1, 2), (3, 4))))
print(check_valid_mx(((1, 2), (3, 4, 5))))
print(check_valid_mx(((1, "a"), (3, 4))))
```

```python
def mx_vec_mult(mx, v):
    """矩阵乘向量，返回结果元组。"""
    if not check_valid_mx(mx):
        raise TypeError("矩阵格式不合法")
    if not isinstance(v, tuple) or not all(isinstance(x, (int, float)) for x in v):
        raise TypeError("向量必须是由整数或浮点数组成的元组")
    if len(mx[0]) != len(v):
        raise ValueError("矩阵列数与向量长度不一致")
    return tuple(sum(a * b for a, b in zip(row, v)) for row in mx)


print(mx_vec_mult(((1, 2), (3, 4)), (5, 6)))
```

## 四、更多向量的乐趣（面向对象）

**维度和元素应作为实例属性，而不是类属性。**

理由：每个向量的维度和元素都属于具体对象，不同对象之间取值不同；
如果做成类属性，所有实例会共享同一份数据，互相覆盖。类属性只适合放所有对象
都相同的常量，因此这里用实例属性 `self.dim`、`self.elements`。

```python
class Vector:
    def __init__(self, dim, elements=None):
        if isinstance(dim, bool) or not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        if elements is None:
            elements = [0] * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(elements) != dim:
            raise ValueError("维度与元素个数不一致")
        if not all(isinstance(x, (int, float)) for x in elements):
            raise TypeError("元素必须是整数或浮点数")
        self.dim = dim
        self.elements = list(elements)

    def _check(self, other):
        if not isinstance(other, Vector):
            raise TypeError("比较对象必须是 Vector")
        if other.dim != self.dim:
            raise ValueError("两个向量维度不同，不能比较")

    def __eq__(self, other):
        self._check(other)
        return self.elements == other.elements

    def __lt__(self, other):
        self._check(other)
        return tuple(self.elements) < tuple(other.elements)

    def __le__(self, other):
        self._check(other)
        return tuple(self.elements) <= tuple(other.elements)

    def __gt__(self, other):
        self._check(other)
        return tuple(self.elements) > tuple(other.elements)

    def dot(self, other):
        self._check(other)
        return float(sum(a * b for a, b in zip(self.elements, other.elements)))

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            return Vector(self.dim, [x * other for x in self.elements])
        if isinstance(other, Vector):
            self._check(other)
            return Vector(self.dim, [a * b for a, b in zip(self.elements, other.elements)])
        raise TypeError("乘法只能与标量或 Vector 进行")

    def __rmul__(self, other):
        return self.__mul__(other)

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"
```

```python
v1 = Vector(2, (2, 3))
v2 = Vector(2, (2, 5))
v3 = Vector(2, (4, 5))

print("v1 < v2:", v1 < v2)
print("v1 == Vector(2, [2, 3]):", v1 == Vector(2, [2, 3]))
print("v2 > v1:", v2 > v1)
print("v1.dot(v3):", v1.dot(v3))
print("2 * v1:", 2 * v1)
print("v1 * 3:", v1 * 3)
print("v1 * v3:", v1 * v3)
```
