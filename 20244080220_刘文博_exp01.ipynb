## 实验2：函数、条件、循环与递归

### 1.编写 score_level(score) 函数：90~100分返回优秀，80~不足90分返回良好，60~不足80分返回及格，0~不足60分返回不及格。超出0~100范围抛出 ValueError。

```python
def score_level(score):
    if not isinstance(score, (int, float)):
        raise TypeError("分数必须是数字")
    if score < 0 or score > 100:
        raise ValueError(f"分数 {score} 超出范围，应为0~100")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

for s in [95, 85, 65, 55, 0, 100]:
    print(f"{s}分 -> {score_level(s)}")

try:
    score_level(105)
except ValueError as e:
    print(f"105分抛出异常：{e}")

try:
    score_level(-10)
except ValueError as e:
    print(f"-10分抛出异常：{e}")
```

### 2.分别使用 for 和 while 计算1到100的整数和，比较两种实现的结果。

```python
sum_for = 0
for i in range(1, 101):
    sum_for += i
print(f"for 循环求和：1+2+...+100 = {sum_for}")

sum_while = 0
n = 1
while n <= 100:
    sum_while += n
    n += 1
print(f"while 循环求和：1+2+...+100 = {sum_while}")
print(f"两种方式结果相等：{sum_for == sum_while}")
```

### 3.编写递归版和迭代版阶乘函数，约定 0! = 1，输入为非负整数。比较 n = 0、1、5、10 时的结果。

```python
def fact_iter(n):
    if n < 0:
        raise ValueError("n必须是非负整数")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

def fact_recur(n):
    if n < 0:
        raise ValueError("n必须是非负整数")
    if n == 0 or n == 1:
        return 1
    return n * fact_recur(n - 1)

for n in [0, 1, 5, 10]:
    fi = fact_iter(n)
    fr = fact_recur(n)
    print(f"{n}! = 迭代版:{fi}  递归版:{fr}  一致:{fi == fr}")
```

### 4.编写 summary(values) 函数，接收非空数值列表，以元组返回最小值、最大值和平均值，并通过元组解包接收结果。

```python
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    mn = min(values)
    mx = max(values)
    avg = sum(values) / len(values)
    return (mn, mx, avg)

data = [85, 58, 92, 76, 60, 89]
min_v, max_v, avg_v = summary(data)
print(f"数据：{data}")
print(f"最小值：{min_v}")
print(f"最大值：{max_v}")
print(f"平均值：{avg_v:.2f}")
```

### 3.编写 parse_student(line) 函数：将成绩转换为整数，检查字段数量、姓名是否为空及成绩是否在0~100范围内。无效记录抛出 ValueError。

```python
def parse_student(line):
    line = line.strip()
    parts = line.split(",")

    if len(parts) != 2:
        raise ValueError(f"字段数量错误：期望2个，实际{len(parts)}个 -> [{line}]")

    name = parts[0].strip()
    score_str = parts[1].strip()

    if not name:
        raise ValueError(f"姓名为空 -> [{line}]")

    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩不是有效整数：[{score_str}] -> [{line}]")

    if score < 0 or score > 100:
        raise ValueError(f"成绩超出0~100范围：{score} -> [{line}]")

    return (name, score)

print("正常记录：", parse_student("Alice,85"))
print("正常记录：", parse_student("Bob,58"))

test_lines = ["Grace,abc", "Henry,105", "Incomplete", ",80", "Eva"]
for tl in test_lines:
    try:
        result = parse_student(tl)
        print(f"  [{tl}] -> {result}")
    except ValueError as e:
        print(f"  [{tl}] -> 异常：{e}")
```

### 4.按行处理文件，捕获无效记录的异常，记录行号与错误原因，并继续处理后续记录。有效记录保存为字典组成的列表。

```python
valid_records = []
invalid_records = []

with open(students_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

for line_no, line in enumerate(lines[1:], start=2):
    line = line.strip()
    if not line:
        continue
    try:
        name, score = parse_student(line)
        valid_records.append({"name": name, "score": score})
    except ValueError as e:
        invalid_records.append({"line_no": line_no, "raw": line, "error": str(e)})

print(f"有效记录数：{len(valid_records)}")
for r in valid_records:
    print(f"  {r}")

print(f"\n无效记录数：{len(invalid_records)}")
for r in invalid_records:
    print(f"  第{r['line_no']}行：[{r['raw']}]  原因：{r['error']}")
```

## 实验5：类、继承与对象复制

### 1.定义 Student 类，在 __init__() 中初始化 name 和 score 属性，并实现 is_passed() 和 __str__() 方法。

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"Student(name={self.name}, score={self.score}, 状态={status})"

s1 = Student("Alice", 85)
s2 = Student("Bob", 58)
print(s1)
print(s2)
print(f"s1.is_passed() = {s1.is_passed()}")
print(f"s2.is_passed() = {s2.is_passed()}")
```

### 3.定义继承自 Student 的 GraduateStudent 类，增加 research_topic 属性，通过 super().__init__() 初始化父类属性，并重写 __str__()。

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"研究方向={self.research_topic}, 状态={status})")

gs1 = GraduateStudent("Carol", 92, "机器学习")
gs2 = GraduateStudent("David", 76, "分布式系统")
print(gs1)
print(gs2)
print(f"gs1.is_passed() = {gs1.is_passed()}  （继承自父类的方法）")
```

### 4.构造包含嵌套列表的对象，例如 name=Alice, scores=[85, 90]。分别进行直接赋值、copy.copy() 和 copy.deepcopy()，修改内部列表，比较结果。每次比较都从新建的原始对象开始。

```python
import copy

original1 = {"name": "Alice", "scores": [85, 90]}
alias1 = original1
alias1["scores"][0] = 999

print("【直接赋值】")
print(f"  original1 = {original1}")
print(f"  alias1    = {alias1}")
print(f"  改 alias 的内部列表，original 也跟着变了 -> 两个是同一个对象\n")

original2 = {"name": "Bob", "scores": [85, 90]}
shallow2 = copy.copy(original2)
shallow2["scores"][0] = 999

print("【浅拷贝 copy.copy()】")
print(f"  original2 = {original2}")
print(f"  shallow2  = {shallow2}")
print(f"  改 shallow 的内部列表，original 还是变了 -> 浅拷贝只复制外层，内层列表还是共享的\n")

original3 = {"name": "Carol", "scores": [85, 90]}
deep3 = copy.deepcopy(original3)
deep3["scores"][0] = 999

print("【深拷贝 copy.deepcopy()】")
print(f"  original3 = {original3}")
print(f"  deep3     = {deep3}")
print(f"  改 deep 的内部列表，original 不变 -> 深拷贝把嵌套的数据也完全复制了一份独立的")
```
