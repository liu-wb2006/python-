## 实验1：正则表达式的基础匹配

```python
import re

text = "cat on mat; later: cats and dogs"
print("\\d+（数字）:", re.findall(r"\d+", text))
print("\\s+（空白）:", re.findall(r"\s+", text))
print("\\bcat\\b（独立单词 cat）:", re.findall(r"\bcat\b", text))
```

```python
# 分支 cat|dog；注意大写 CAT 不会被匹配
print("cat|dog:", re.findall(r"cat|dog", "cat dog bird CAT"))

# 分支按从左到右尝试，顺序不同结果不同
print(r"a|a|a|aaa :", re.findall(r"a|a|a|aaa", "aaaa"))
print(r"aaa|a|a   :", re.findall(r"aaa|a|a", "aaaa"))
```

`a|a|a|aaa` 中每个位置先尝试单个 `a`，永远走不到 `aaa`；
`aaa|a|a` 先尝试三个 `a`，所以能匹配出 `aaa` 再加一个 `a`。分支顺序会影响匹配结果。

```python
samples = "ct cat caat caaat"
for pat in [r"ca*t", r"ca+t", r"ca?t", r"ca{2}t", r"ca{1,2}t"]:
    print(pat, "->", re.findall(pat, samples))
```

- `ca*t`：`a` 出现 0 次或多次；
- `ca+t`：`a` 出现 1 次或多次；
- `ca?t`：`a` 出现 0 次或 1 次；
- `ca{2}t`：`a` 恰好出现 2 次；
- `ca{1,2}t`：`a` 出现 1 到 2 次。

## 实验2：分组、回溯引用与替换

```python
# 1. 提取邮箱的用户名和域名
emails = "联系 : alice@example.com ; 备选 : bob@test.org"
for user, domain in re.findall(r"(\w+)@(\w+\.\w+)", emails):
    print("用户名:", user, " 域名:", domain)
```

```python
# 2. 后向引用找出连续重复的单词
words = "cat cat, dog cat, bird bird"
print("重复单词:", re.findall(r"\b(\w+)\s+\1\b", words))
```

```python
# 3. 日期 年-月-日 转 月-日-年，保留行数
dates = "2024-05-01\n2024/05/02\n2024.05.03"
converted = re.sub(r"(\d{4})[-/.](\d{2})[-/.](\d{2})", r"\2-\3-\1", dates)
print("转换前:\n", dates)
print("转换后:\n", converted)
```

```python
# 4. 电话号码统一为 (123) 456-7890 形式
phones = "联系我：123-456-7890 或 987.654.3210"
formatted = re.sub(r"(\d{3})[-.](\d{3})[-.](\d{4})", r"(\1) \2-\3", phones)
print(formatted)
```

## 实验3：使用 BeautifulSoup 解析 HTML

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")
print("页面标题文本:", soup.title.string)
print("h1 标签名:", soup.h1.name, " h1 文本:", soup.h1.get_text())
```

```python
# 所有 class="course" 的段落
for p in soup.find_all("p", class_="course"):
    print("段落:", p.get_text())

# 所有链接：文字与 href
for a in soup.find_all("a"):
    print("链接文字:", a.get_text(), " href:", a["href"])
```

```python
# get_text() 提取可见文本
print("整页可见文本:\n", soup.get_text())
print("单个段落文本:", soup.find("p", class_="course").get_text())
print("单个链接文本:", soup.find("a").get_text())
```
